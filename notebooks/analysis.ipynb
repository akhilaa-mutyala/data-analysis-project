```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
print("library loaded")
```

```python
df=pd.read_csv("C:\\Users\\Akhila\\Downloads\\titanic.zip")
df.head()
```

```python
print("rows and columns:",df.shape)
```

```python
print(df.isnull().sum())
```

```python
# Fill missing Age values
df["Age"]=df["Age"].fillna(df["Age"].mean())
# Safely drop Cabin if it exists
df.drop(columns=["Cabin"], inplace=True, errors="ignore")
print("Cleaned! ")
```

```python
sns.countplot(x="Survived", data=df)
plt.title("How many survived?   1=survive, 0=dead")
plt.savefig("survival.png")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.savefig("heatmap.png")
plt.show()
print(" saved!")
```

```python
import urllib.request
import os

# Create data folder one level up
os.makedirs("../data", exist_ok=True)

# Download titanic dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
urllib.request.urlretrieve(url, "../data/titanic.csv")

print("Downloaded! ")
```

```python
import pandas as pd
df = pd.read_csv("../data/titanic.csv")
print("Rows:", len(df))
df.head()
```

```python
import os

df["Age"] = df["Age"].fillna(df["Age"].mean())
df.drop(columns=["Cabin"], inplace=True, errors="ignore")
os.makedirs("../outputs", exist_ok=True)

print("Cleaned! ")
```

```python
sns.countplot(x="Sex", hue="Survived", data=df)
plt.title("Survival by Gender")
plt.savefig("../outputs/gender.png")
plt.show()
```

```python
sns.histplot(df["Age"], bins=30, kde=True)
plt.title("Age Distribution")
plt.savefig("../outputs/age.png")
plt.show()
```

```python
sns.countplot(x="Pclass", hue="Survived", data=df)
plt.title("Survival by Class")
plt.savefig("../outputs/class.png")
plt.show()
```
